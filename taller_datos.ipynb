## Ejercicio 1: Operaciones fundamentales con NumPy

### 1.1 Arreglo 1D: forma y tipo de dato

```python
import numpy as np
import pandas as pd

arr1d = np.arange(1, 13)
print(arr1d)
print('shape:', arr1d.shape)
print('dtype:', arr1d.dtype)
```

### 1.2 Cambio de forma con reshape()

```python
arr2d = arr1d.reshape(3, 4)
print(arr2d)
print('shape:', arr2d.shape)
```

### 1.3 Matrices aleatorias

```python
np.random.seed(42)
alea_rand = np.random.rand(3, 3)      # flotantes en [0, 1)
alea_int = np.random.randint(0, 10, size=(3, 3))  # enteros en [0, 10)
print(alea_rand)
print(alea_int)
```

### 1.4 Matrices estructurales

```python
print(np.zeros((2, 3)))
print(np.ones((2, 3)))
print(np.eye(3))
```

### 1.5 Transpuesta (.T)

```python
print(alea_rand)
print(alea_rand.T)
```

### 1.6 Concatenación horizontal y vertical

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.hstack((a, b)))
print(np.vstack((a, b)))
```

### 1.7 Operaciones matemáticas

```python
x = np.array([4, 8, 15, 16, 23, 42])
print('suma:', np.sum(x))
print('promedio:', np.mean(x))
print('mediana:', np.median(x))
print('producto:', np.prod(x))
print('desv. estándar:', np.std(x))
print('varianza:', np.var(x))
print('mínimo:', np.min(x))
print('máximo:', np.max(x))
print('suma acumulativa:', np.cumsum(x))
```

### 1.8 Operaciones elemento a elemento

```python
u = np.array([1, 2, 3, 4])
v = np.array([10, 20, 30, 40])
print('suma:', u + v)
print('resta:', v - u)
print('multiplicación:', u * v)
```

## Ejercicio 2: Integración y manipulación con Pandas

### 2.1 Importación con alias estándar
Ya lo realize en la primera celda: `import pandas as pd` e `import numpy as np`.

### 2.2 DataFrame desde un arreglo NumPy 2D

```python
datos_np = np.random.randint(1, 100, size=(4, 3))
df_np = pd.DataFrame(datos_np, columns=['A', 'B', 'C'])
df_np
```

### 2.3 DataFrame desde un diccionario

```python
diccionario = {
    'Nombre': ['Ana', 'Luis', 'Marta', 'Pedro'],
    'Edad': [23, 31, 28, 35],
    'Nota': [4.2, 3.8, 4.5, 3.9]
}
df_dict = pd.DataFrame(diccionario)
df_dict
```

### 2.4 DataFrame a arreglo NumPy

```python
arr_a = df_np.to_numpy()
arr_b = df_np.values
print(arr_a)
print(type(arr_a), type(arr_b))
print('Iguales:', np.array_equal(arr_a, arr_b))
```

### 2.5 Promedios con np.mean() sobre el DataFrame

```python
print('Por columna (axis=0):')
print(np.mean(df_np, axis=0))
print('Por fila (axis=1):')
print(np.mean(df_np, axis=1))
```

### 2.6 Promedios con métodos nativos de Pandas

```python
print('Por columna (axis=0):')
print(df_np.mean(axis=0))
print('Por fila (axis=1):')
print(df_np.mean(axis=1))
```
